# 📱 Google Play Store EDA Project

## 🧾 Overview

This project performs **Hypothesis Testing** on a dataset containing information about over **2.3 Million Android applications** listed on the **Google Play Store**. The goal is to Test Hypothesis that
Testable Hypothesis: Relationship Between App Category, Pricing Model, and User Engagement
Proposed Hypothesis:
"Apps in high-demand categories (e.g., Tools, Communication, Productivity) that adopt a freemium model (free with in-app purchases) will exhibit higher average user ratings and install counts compared to fully paid apps in the same categories, due to broader accessibility and monetization flexibility."
Rationale from Provided Insights:

High-demand categories (from Sections 2 & 3): Tools, Communication, and Productivity show the highest installs and ratings, indicating strong user engagement and daily utility.
Pricing dominance (from Section 4): 98% of apps are free, suggesting paid apps face barriers to adoption; freemium (leveraging Free=True + In-App Purchases=True) aligns with this trend for revenue without alienating users.
Monetization strategy (from Section 6 & Recommendations): Categories with lower average prices benefit from in-app purchases, and high-value categories justify premiums only if engagement is proven—testing this hypothesis validates freemium as a low-risk entry for developers targeting crowded, high-engagement spaces.

Why This is Hypothesis:
Using columns like Category, Free, In-App Purchases, Rating, Rating Count, Installs (or Minimum/Maximum Installs for estimates), and Price:

### **Null Hypothesis (H0)**: No significant difference in average Rating or Installs between freemium apps and fully paid apps within high-demand categories.
### **Alternative Hypothesis (H1)**: Freemium apps have significantly higher average Rating (>0.2 points) and Installs (>10x) than fully paid apps in these categories..

> #### 3.2.2 Clean the `Size` column.

```python
# check the unique values
df['Size'].unique()
```

```python
# By checking value counts will help to explore the more unique values
df['Size'].value_counts().head()
```

### Solution:
To prepare the `Size` column for analysis, we must:
1. Convert the unit of measurement (M,G or K) to a consistent format, such as bytes.
2. Remove the unit suffix ("M","G" or "K") from the values.
3. Change the data type of the values to numeric.

```python
df['Size'].isnull().sum()
```

```python
# Remove commas from values
df['Size'] = df['Size'].str.replace(',', '')
```

> Let's Convert "M","G" and "k" units in storage values to bytes by removing the respective letters and converting the remaining numeric data to a numeric data type.

```python
# Change the "size" column to a numeric format by performing the following transformations:
#  1. If the value contains the letter "k", multiply it by 1024.
#  2. If the value contains the letter "M", multiply it by 1024*1024.
#  2. If the value contains the letter "G", multiply it by 1024*1024*1024.
#  3.this function will convert the size column to numeric
def convert_size(size):
    # add function details here
    '''
    This function transforms the "size" column into a numeric format.
    It multiplies the values by 1024 if there is a "k" in the value, and by 1024*1024 if there is an "M" in the value.
    '''
def convert_size(size):
    if isinstance(size, str):
        size = size.replace(',', '')  # Remove commas from the value
        if 'k' in size:
            return float(size.replace('k', '')) * 1024
        elif 'M' in size:
            return float(size.replace('M', '')) * 1024 * 1024
        elif 'G' in size:
            return float(size.replace('G', '')) * 1024 * 1024 * 1024
        elif 'Varies with device' in size:
            return np.nan
    return size
# Apply the function to the Size
df['Size'] = df['Size'].apply(convert_size)
```

```python
# Let's rename the Size column to Size_in_bytes
df.rename(columns={'Size': 'Size_in_bytes'}, inplace=True)
```

```python
# making a new column called 'Size in Mb' which will have the size in MB
df['Size_in_Mb'] = df['Size_in_bytes'].apply(lambda x: x/(1024*1024))
```

- Each value has been converted to its byte equivalent, with the `"M"` ,`"G"` and `"K"` suffixes removed.
- The resulting values are all numeric data types.
- The `"Varies with device"` value was originally a string.
- For data consistency, we have temporarily replaced it with a null value. We can assign appropriate values later as needed.

## **10. Hypothesis Testing Using ANOVA and Chi-Squared test**

```python

from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
from scipy.stats import chi2_contingency

# Define high-demand categories
high_demand_categories = ['Tools', 'Communication', 'Productivity']

# Filter test_df (assuming 'df' is your main DataFrame)
test_df = df[
    df['Category'].isin(high_demand_categories) &
    df['Rating'].notnull() &
    df['Installs'].notnull() &
    (df['Type'] == 'Paid')
].copy()

# Data Preparation
test_df['Pricing_Model'] = np.where(test_df['In App Purchases'] == True, 'Freemium', 'Fully Paid')
test_df['Installs_Log10'] = np.log10(test_df['Installs'] + 1)

# FIX: Convert to legacy float64 to resolve nullable dtype error
for col in ['Rating', 'Installs', 'Installs_Log10']:
    if col in test_df.columns:
        test_df[col] = test_df[col].astype('float64')
print("✅ Converted numeric columns to float64.")

# Make categorical
test_df['Pricing_Model'] = pd.Categorical(test_df['Pricing_Model'])

# Create Installs_category for Test 3 (if missing)
if 'Installs_category' not in test_df.columns:
    def categorize_installs(installs):
        if installs == 0: return 'no'
        elif installs < 1000: return 'Very low'
        elif installs < 10000: return 'Low'
        elif installs < 100000: return 'Moderate'
        elif installs < 1000000: return 'Good'
        elif installs < 10000000: return 'High'
        elif installs < 100000000: return 'Very High'
        else: return 'Top Notch'
    test_df['Installs_category'] = test_df['Installs'].apply(categorize_installs)

# Define groups
freemium_group = test_df[test_df['Pricing_Model'] == 'Freemium']
fully_paid_group = test_df[test_df['Pricing_Model'] == 'Fully Paid']

print(f"Freemium Group Size: {len(freemium_group)}")
print(f"Fully Paid Group Size: {len(fully_paid_group)}")
print("-" * 50)

if len(freemium_group) < 2 or len(fully_paid_group) < 2:
    print("⚠️ WARNING: Not enough data for robust testing.")

alpha = 0.05

# Test 1: ANOVA for Rating
print("### Test 1: One-Way ANOVA for Average Rating ###")
if len(freemium_group) >= 2 and len(fully_paid_group) >= 2:
    try:
        model_rating = ols('Rating ~ C(Pricing_Model)', data=test_df).fit()
        anova_table_rating = anova_lm(model_rating)
        p_rating = anova_table_rating['PR(>F)'][0]
        
        freemium_avg_rating = freemium_group['Rating'].mean()
        fully_paid_avg_rating = fully_paid_group['Rating'].mean()
        diff_rating = freemium_avg_rating - fully_paid_avg_rating
        required_diff = 0.2
        
        print(f"Freemium Avg Rating: {freemium_avg_rating:.3f}")
        print(f"Fully Paid Avg Rating: {fully_paid_avg_rating:.3f}")
        print(f"ANOVA P-value: {p_rating:.4e}")
        print(f"Difference (Freemium - Fully Paid): {diff_rating:.4f}")
        
        if p_rating < alpha and abs(diff_rating) > required_diff:
            print("✅ Conclusion: Reject H0. Significant and meets magnitude threshold.")
        elif p_rating < alpha:
            print("🔶 Conclusion: Reject H0. Significant but magnitude insufficient.")
        else:
            print("❌ Conclusion: Fail to Reject H0. No significant difference.")
    except Exception as e:
        print(f"❌ Error in Test 1: {e}")
else:
    print("Not enough data for Test 1.")

# Test 2: ANOVA for Log-Transformed Installs
print("\n### Test 2: One-Way ANOVA for Log-Transformed Installs ###")
if len(freemium_group) >= 2 and len(fully_paid_group) >= 2:
    try:
        model_installs = ols('Installs_Log10 ~ C(Pricing_Model)', data=test_df).fit()
        anova_table_installs = anova_lm(model_installs)
        p_installs = anova_table_installs['PR(>F)'][0]
        
        freemium_avg_installs = freemium_group['Installs'].mean()
        fully_paid_avg_installs = fully_paid_group['Installs'].mean()
        installs_ratio = freemium_avg_installs / fully_paid_avg_installs if fully_paid_avg_installs > 0 else np.inf
        log_diff = freemium_group['Installs_Log10'].mean() - fully_paid_group['Installs_Log10'].mean()
        required_log_diff = np.log10(10)  # ~1.0 for 10x
        
        print(f"Freemium Avg Installs: {freemium_avg_installs:,.0f}")
        print(f"Fully Paid Avg Installs: {fully_paid_avg_installs:,.0f}")
        print(f"ANOVA P-value: {p_installs:.4e}")
        print(f"Installs Ratio (Freemium / Fully Paid): {installs_ratio:.2f}x")
        print(f"Log10 Difference: {log_diff:.4f} (Threshold: >{required_log_diff:.1f} for 10x)")
        
        if p_installs < alpha and log_diff > required_log_diff:
            print("✅ Conclusion: Reject H0. Significant and meets 10x magnitude.")
        elif p_installs < alpha:
            print("🔶 Conclusion: Reject H0. Significant but <10x magnitude.")
        else:
            print("❌ Conclusion: Fail to Reject H0. No significant difference.")
    except Exception as e:
        print(f"❌ Error in Test 2: {e}")
else:
    print("Not enough data for Test 2.")

# Test 3: Chi-Square for Pricing vs. Installs Category (FIXED)
print("\n### Test 3: Chi-Square Test for Pricing Model vs. Installs Category ###")
contingency_table = pd.crosstab(test_df['Pricing_Model'], test_df['Installs_category'])

expected_categories = ['no', 'Very low', 'Low', 'Moderate', 'Good', 'High', 'Very High', 'Top Notch']
contingency_table = contingency_table.reindex(columns=expected_categories, fill_value=0)

print("Full Contingency Table:\n", contingency_table)

# FIX: Filter to categories with >0 in BOTH groups to avoid zero expected frequencies
valid_cats = contingency_table.columns[
    (contingency_table.loc['Freemium', :] > 0) & 
    (contingency_table.loc['Fully Paid', :] > 0)
]

if len(valid_cats) > 1:  # Need at least 2 categories for meaningful test
    contingency_filtered = contingency_table[valid_cats].copy()
    print(f"\nFiltered Categories (with variation in both groups): {list(valid_cats)}")
    print("Filtered Contingency Table:\n", contingency_filtered)
    
    try:
        chi2, p_chi2, dof, expected = chi2_contingency(contingency_filtered)
        print(f"\nChi-Square Statistic: {chi2:.4f}, DoF: {dof}")
        print(f"Chi-Square P-value: {p_chi2:.4e}")
        
        # Warn on low expected
        if np.any(expected < 5):
            print("⚠️ Some expected frequencies <5; results approximate for small cells.")
        
        if p_chi2 < alpha:
            print("✅ Conclusion: Reject H0. Significant association between Pricing Model and Installs Category.")
        else:
            print("❌ Conclusion: Fail to Reject H0. No significant association.")
    except Exception as e:
        print(f"❌ Still an error after filtering: {e}")
else:
    print(f"⚠️ Too few categories with variation in both groups ({len(valid_cats)}). Cannot run Chi-Square—data too sparse for association test.")
    print("Recommendation: Combine rare categories (e.g., 'Very High' + 'Top Notch' into 'Elite') or use a different test like logistic regression.")
```
